```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("insurance.csv")
df
```

```python
df.columns
```

```python

plt.bar(df["region"], df["charges"])

plt.xlabel("Region")
plt.ylabel("Charges")
plt.title("Insurance Charges by Region")

plt.show()
```

```python
plt.scatter(df["age"],df["bmi"])
plt.xlabel("AGE")
plt.ylabel("BMI")
plt.title("Age VS BMI")
plt.show()
```

```python
sns.regplot(x="age", y="bmi", data=df)

plt.title("Age vs BMI Regression")
plt.show()
```

```python
sns.countplot(x="smoker", data=df)

plt.show()
```

```python
sns.countplot(x=df["sex"], data= df)
plt.show()
```

```python
sns.histplot(data=df, x="children", kde=True, bins=20,discrete=True)

plt.show()
```

```python
region_count = df["region"].value_counts()

plt.figure(figsize=(8,6))

plt.pie(region_count.values,
        labels=region_count.index,
        autopct="%1.1f%%",
        startangle=90,
        shadow=True)

plt.title("Region Distribution")

plt.show()
```
